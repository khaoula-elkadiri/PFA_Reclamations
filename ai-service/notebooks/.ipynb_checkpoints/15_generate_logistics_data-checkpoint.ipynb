# Notebook 15 — Génération Dataset Logistique Français

**Objectif :** Créer un dataset de réclamations logistiques réalistes en français, via templates avec variations de vocabulaire.

**Problème résolu :** L'ancien dataset venait de réclamations financières américaines traduites automatiquement → le modèle ne comprenait pas le vocabulaire logistique français.

**Résultat attendu :** ~200 exemples par catégorie = ~1400 exemples de haute qualité.

```python
import pandas as pd
import random
import itertools
from pathlib import Path

random.seed(42)

DATA_OUT = Path('../data/processed/reclamations_logistics_fr.csv')
```

## 1. Fonctions utilitaires

```python
def generer(templates, variations, n=200):
    """Génère n exemples en remplissant les templates avec les variations."""
    exemples = set()
    tentatives = 0
    while len(exemples) < n and tentatives < n * 10:
        t = random.choice(templates)
        remplacement = {k: random.choice(v) for k, v in variations.items()}
        try:
            texte = t.format(**remplacement)
            exemples.add(texte)
        except KeyError:
            pass
        tentatives += 1
    return list(exemples)[:n]
```

## 9. Assemblage du dataset final

```python
# Assembler tous les exemples
all_data = []

categories_data = [
    (retard_examples, "retard_livraison"),
    (casse_examples, "produit_casse"),
    (picking_examples, "erreur_picking"),
    (manquant_examples, "article_manquant"),
    (qualite_examples, "mauvaise_qualite"),
    (transport_examples, "probleme_transport"),
    (admin_examples, "erreur_administrative"),
]

for examples, categorie in categories_data:
    for texte in examples:
        all_data.append({"texte": texte, "categorie": categorie})

df_new = pd.DataFrame(all_data)

# Mélanger
df_new = df_new.sample(frac=1, random_state=42).reset_index(drop=True)

print(f"\nDistribution des catégories (nouveau dataset):")
print(df_new['categorie'].value_counts())
print(f"\nTotal: {len(df_new)} exemples")
```

```python
# Charger l'ancien dataset et fusionner
df_ancien = pd.read_csv('../data/processed/reclamations_clean_augmente.csv')
print(f"Ancien dataset: {len(df_ancien)} exemples")
print(df_ancien.columns.tolist())

# L'ancien dataset a des colonnes texte_clean et categorie
# Renommer pour correspondre
df_ancien_norm = df_ancien[['texte_clean', 'categorie']].rename(columns={'texte_clean': 'texte'})

# Combiner les deux datasets
df_combined = pd.concat([df_new, df_ancien_norm], ignore_index=True)
df_combined = df_combined.drop_duplicates(subset=['texte']).reset_index(drop=True)

print(f"\nDataset combiné: {len(df_combined)} exemples")
print("\nDistribution finale:")
print(df_combined['categorie'].value_counts())
```

```python
# Sauvegarder
df_combined.to_csv(DATA_OUT, index=False)
print(f"Dataset sauvegardé : {DATA_OUT}")
print(f"Colonnes : {df_combined.columns.tolist()}")
print("\n5 premiers exemples :")
df_combined.head()
```
