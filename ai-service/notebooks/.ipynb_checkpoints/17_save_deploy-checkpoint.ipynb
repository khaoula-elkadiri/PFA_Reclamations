# Notebook 17 — Sauvegarde et Déploiement du Modèle Final

**Objectif :** Entraîner le meilleur pipeline sur TOUT le dataset (train + test) et sauvegarder dans `models/`.

**Workflow :**
1. Charger le dataset complet (`reclamations_logistics_fr.csv`)
2. Entraîner LinearSVC calibré sur 100% des données
3. Sauvegarder `models/classifier_pipeline.pkl` et `models/categories.pkl`
4. Test rapide de sanité sur le modèle sauvegardé

```python
import pandas as pd
import numpy as np
import joblib
import sys, os
sys.path.insert(0, os.path.abspath('..'))

from sklearn.pipeline import Pipeline
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.svm import LinearSVC
from sklearn.calibration import CalibratedClassifierCV
from sklearn.model_selection import cross_val_score, StratifiedKFold
from sklearn.metrics import accuracy_score, classification_report
from pathlib import Path

DATA_PATH = Path('../data/processed/reclamations_logistics_fr.csv')
MODEL_PATH = Path('../models/classifier_pipeline.pkl')
CATEGORIES_PATH = Path('../models/categories.pkl')

print("Packages chargés OK")
```

## 1. Chargement du dataset complet

```python
df = pd.read_csv(DATA_PATH)
df = df.dropna(subset=['texte', 'categorie'])
df = df[df['texte'].str.strip().str.len() > 0]

X = df['texte'].values
y = df['categorie'].values
categories = sorted(df['categorie'].unique().tolist())

print(f"Dataset total : {len(df)} exemples")
print(f"Catégories ({len(categories)}) : {categories}")
print("\nDistribution :")
print(df['categorie'].value_counts())
```

## 2. Validation croisée finale (estimation de performance réelle)

```python
# Pipeline final
pipeline_final = Pipeline([
    ('tfidf', TfidfVectorizer(
        max_features=15000,
        ngram_range=(1, 3),
        sublinear_tf=True,
        min_df=1,
        max_df=0.95,
        analyzer='word',
        strip_accents='unicode',
        lowercase=True,
    )),
    ('clf', CalibratedClassifierCV(
        LinearSVC(C=1.0, max_iter=5000, class_weight='balanced'),
        cv=5
    ))
])

# Validation croisée 5-fold pour estimer la performance réelle
print("Validation croisée 5-fold en cours...")
cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
scores = cross_val_score(pipeline_final, X, y, cv=cv, scoring='accuracy', n_jobs=-1)

print(f"\nScores CV : {[f'{s:.2%}' for s in scores]}")
print(f"Accuracy CV : {scores.mean():.2%} ± {scores.std():.2%}")
print(f"(Ancien modèle : ~78% accuracy)")
```

## 3. Entraînement sur 100% du dataset

```python
print("Entraînement final sur 100% du dataset...")
pipeline_final.fit(X, y)
print("Entraînement terminé.")

# Vérification rapide sur les données d'entraînement (doit être proche de 100%)
y_pred_train = pipeline_final.predict(X)
print(f"Accuracy sur train (overfit check) : {accuracy_score(y, y_pred_train):.2%}")
```

## 4. Sauvegarde

```python
# Sauvegarder le pipeline et les catégories
joblib.dump(pipeline_final, MODEL_PATH)
joblib.dump(categories, CATEGORIES_PATH)

print(f"Modèle sauvegardé : {MODEL_PATH}")
print(f"Catégories sauvegardées : {CATEGORIES_PATH}")
print(f"Taille du modèle : {MODEL_PATH.stat().st_size / 1024:.1f} KB")
```

## 5. Test de sanité du modèle sauvegardé

```python
# Recharger et tester (simule ce que fait l'API au démarrage)
modele_charge = joblib.load(MODEL_PATH)
categories_chargees = joblib.load(CATEGORIES_PATH)

print(f"Modèle rechargé OK")
print(f"Classes du modèle : {list(modele_charge.classes_)}")
print(f"Catégories fichier : {categories_chargees}")
assert set(modele_charge.classes_) == set(categories_chargees), "ERREUR: Discordance catégories!"
print("Catégories cohérentes ✓")
```

```python
# Tests de sanité sur des cas représentatifs
tests_sanite = [
    ("Ma commande n'est pas arrivée depuis 10 jours.", "retard_livraison"),
    ("J'ai reçu mon téléphone cassé, l'écran est brisé.", "produit_casse"),
    ("J'ai commandé du bleu et reçu du rouge, mauvais article.", "erreur_picking"),
    ("Il manque le chargeur dans mon colis.", "article_manquant"),
    ("Produit de très mauvaise qualité, tombé en panne après 2 jours.", "mauvaise_qualite"),
    ("Le transporteur a livré mon colis à la mauvaise adresse.", "probleme_transport"),
    ("J'ai été débité deux fois pour la même commande.", "erreur_administrative"),
]

correct = 0
print(f"{'Texte':<55} {'Attendu':<22} {'Prédit':<22} {'Conf':>6} {'OK?'}")
print("-" * 115)

for texte, attendu in tests_sanite:
    pred = modele_charge.predict([texte])[0]
    proba = modele_charge.predict_proba([texte])[0]
    conf = max(proba)
    ok = pred == attendu
    if ok:
        correct += 1
    print(f"{texte[:53]:<55} {attendu:<22} {pred:<22} {conf:>5.0%} {'✓' if ok else '✗'}")

print(f"\nRésultat sanité : {correct}/{len(tests_sanite)} correctes")

if correct >= 6:
    print("Modèle validé et prêt pour l'API ✓")
else:
    print("ATTENTION : performance insuffisante, vérifier le dataset")
```
