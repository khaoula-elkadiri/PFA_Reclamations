```python
# Tous les nouveaux exemples
tous_nouveaux = (
    nouveaux_exemples_manquant + 
    nouveaux_exemples_picking + 
    nouveaux_exemples_retard
)

# Convertir en DataFrame
df_nouveaux = pd.DataFrame(tous_nouveaux, columns=['texte_clean', 'categorie', 'priorite'])
print(f"Total nouveaux exemples : {len(df_nouveaux)}")

# Appliquer le preprocessing aux nouveaux exemples
import spacy
import unicodedata
import re

nlp = spacy.load("fr_core_news_md")

def preprocesser(texte):
    texte = str(texte).lower().strip()
    texte = unicodedata.normalize('NFKD', texte)
    texte = ''.join([c for c in texte if not unicodedata.combining(c)])
    texte = re.sub(r'[^\w\s]', ' ', texte)
    texte = re.sub(r'\d+', ' ', texte)
    texte = re.sub(r'\s+', ' ', texte).strip()
    doc = nlp(texte)
    # MODIFICATION : on garde les mots à partir de 2 caractères pour ne pas perdre "pas", "non"
    tokens = [t.lemma_ for t in doc 
              if not t.is_stop and not t.is_punct 
              and len(t.text) >= 2 and t.is_alpha]
    return ' '.join(tokens)

df_nouveaux['texte_clean'] = df_nouveaux['texte_clean'].apply(preprocesser)

# Filtrer les textes vides après preprocessing
df_nouveaux = df_nouveaux[df_nouveaux['texte_clean'].str.split().str.len() >= 2]
print(f"Après preprocessing : {len(df_nouveaux)} exemples valides")

# Fusionner avec l'ancien dataset
df_final = pd.concat([df_existant, df_nouveaux], ignore_index=True)
df_final = df_final.drop_duplicates(subset=['texte_clean'])

print(f"\nDataset final : {len(df_final)} réclamations")
print("\nNouvelle distribution :")
print(df_final['categorie'].value_counts())

# Sauvegarder
df_final.to_csv('../data/processed/reclamations_clean_augmente.csv', index=False)
print("\n✅ Dataset augmenté sauvegardé !")
```
